**Conceptual Questions:**

**Suppose we obtain a bootstrap sample from a set of n observations. Sampling is with replacement.**

**1. What is the probability that your first bootstrap observation is not in the _jth_ observation?**

We have $n$ observations. To get the first bootstrap observation, we pick one of them at random.

**It is easier to first find the probability that the draw *is* the $j$th observation.**

Only 1 of the $n$ observations is the $j$th one, so:

$$P(\text{first bootstrap observation} = j\text{th observation}) = \frac{1}{n}$$

**We need the probability that it is *not* the $j$th observation.**

"Not the $j$th" is the opposite (complement) of "is the $j$th", so the two probabilities add to 1:

$$P(\text{first bootstrap observation} \neq j\text{th observation}) = 1 - \frac{1}{n} = \frac{n-1}{n}$$

This is true, because out of $n$ equally likely choices, $n-1$ of them are *not* the $j$th observation, so the chance of landing on one of those is $\frac{n-1}{n}$.

**2. What is the probability that your second bootstrap point is in the _jth_ observation?**

Bootstrap sampling is *with replacement*, so whatever we picked  first is put back. The second draw again chooses from all $n$ original observations, each equally likely.

Then we need to look at the probability that the second draw is the $j$th observation.

$$P(\text{second bootstrap observation} = j\text{th observation}) = \frac{1}{n}$$

From previous logic, we can ascertain that the first draw has no influence on the probabilities for the second draw, they are independent. This is expected for sampling with replacement, and the key difference between with and without replacement.


**3. How often the _ith_ observation is excluded from the _jth_ sample? Are these two mutually exclusive?**

From the earlier questions, the chance that any single draw is *not* the $j$th observation is:

$$P(\text{one draw misses } j) = 1 - \frac{1}{n}$$

A whole bootstrap sample has $n$ draws. For the $i$th observation to be excluded, every one of the $n$ draws must miss it. The draws are independent, so we multiply the probabilities:

$$P(i\text{th observation not in the bootstrap sample}) = \underbrace{\left(1 - \frac{1}{n}\right) \times \left(1 - \frac{1}{n}\right) \times \cdots \times \left(1 - \frac{1}{n}\right)}_{n \text{ times}} = \left(1 - \frac{1}{n}\right)^n$$

$$n = 5: \quad \left(1 - \tfrac{1}{5}\right)^5 = 0.8^5 \approx 0.328$$

$$n = 10: \quad \left(1 - \tfrac{1}{10}\right)^{10} = 0.9^{10} \approx 0.349$$

$$n = 100: \quad \left(1 - \tfrac{1}{100}\right)^{100} = 0.99^{100} \approx 0.366$$

As $n$ grows, this settles at a fixed value:

$$\lim_{n \to \infty} \left(1 - \frac{1}{n}\right)^n = e^{-1} \approx 0.368$$

**So "how often?"** if you create many bootstrap samples, a given observation is left out about $36.8\%$ of them, and appears in about $63.2\%$.

**Mutually exclusive?**

Two events $A$ and $B$ are *mutually exclusive* if they cannot both happen, meaning $P(A \text{ and } B) = 0$.

Let $A$ = "draw 1 is not the $j$th observation" and $B$ = "draw 1 is not the the $j$th observation". Both can happen at once, so they are **not** mutually exclusive:

$$P(A \text{ and } B) = \left(1 - \frac{1}{n}\right)^2 > 0$$

Independence and mutual exclusivity are different ideas. The events that *are* mutually exclusive are "the $j$th observation is in the sample" and "the $j$th observation is not in the sample". They can't both happen.


**4. How does the probability of the bootstrap being in the _jth_ observation change if we have a tightly constrained dataset?**

The chance that the $j$th observation is *left out* of a bootstrap sample of $n$ draws is:

$$P(\text{not in sample}) = \left(1 - \frac{1}{n}\right)^n$$

So the chance that it is *in* the sample is the complement:

$$P(\text{in sample}) = 1 - \left(1 - \frac{1}{n}\right)^n$$

I think we can gather from our  equations in the previous question, that the smaller the dataset, the *more likely*  any one observation is to appear in a bootstrap sample. The probability drops toward $0.632$ as $n$ grows.

**5. How could the probability of the bootstrap being in the _jth_ observation change if we have 20 times that size?**

The formula is the same, but with $20n$ observations:

$$P(\text{in sample}) = 1 - \left(1 - \frac{1}{20n}\right)^{20n}$$

Comparing small n (2, 3, 5) with $20n$.

$$n = 2 \to 40: \quad 1 - \left(\tfrac{1}{2}\right)^2 = 1 - 0.25 = 0.750 \;\to\; 1 - \left(\tfrac{39}{40}\right)^{40} \approx 0.637$$

$$n = 3 \to 60: \quad 1 - \left(\tfrac{2}{3}\right)^3 = 1 - 0.296 \approx 0.704 \;\to\; 1 - \left(\tfrac{59}{60}\right)^{60} \approx 0.635$$

$$n = 5 \to 100: \quad 1 - (0.8)^5 = 1 - 0.328 \approx 0.672 \;\to\; 1 - (0.99)^{100} \approx 0.634$$

Thus, with a 20x increase from a small dataset, the probability *decreases* slightly and moves toward its limit:

$$1 - \left(1 - \frac{1}{20n}\right)^{20n} \;\longrightarrow\; 1 - e^{-1} \approx 0.632$$

The reason it changes so little is each draw is 20 times less likely to pick the $j$th one, but there are also 20 times as many draws (basically cancelling out).


**How is the _jth_ observation's probability altered if our dataset is considered to be 'big data'? How would this fact alter each sample?**

"Big data" means $n$ is very large, so this is the limiting case from the last two answers. Here the probability stops depending on $n$. Rather, for very large n, the formula settles at a fixed value:

$$P(j\text{th observation not in sample}) = \left(1 - \frac{1}{n}\right)^n \;\longrightarrow\; e^{-1} \approx 0.368$$

$$P(j\text{th observation in sample}) = 1 - \left(1 - \frac{1}{n}\right)^n \;\longrightarrow\; 1 - e^{-1} \approx 0.632$$

Example with $n = 1{,}000{,}000$: $(0.999999)^{1{,}000{,}000} \approx 0.3679$, so the chance of being in the sample is about $0.6321$. It is basically the limit.

What this means for each bootstrap sample is that each observation has about a $63.2\%$ chance of being included, so on average a sample contains:

$$\text{unique observations} \approx 0.632\,n \qquad \text{left out} \approx 0.368\,n$$
